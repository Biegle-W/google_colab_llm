### Keep downloaded models on Google Drive (so they are not re-downloaded every session)

```python
from google.colab import drive
import os
drive.mount("/content/drive")
HF_CACHE = "/content/drive/MyDrive/llm_models/hf_cache"   # models are saved here and reused next session
os.makedirs(HF_CACHE, exist_ok=True)
os.environ["HF_HOME"] = HF_CACHE
os.environ["HF_HUB_CACHE"] = HF_CACHE + "/hub"
print("Hugging Face cache:", HF_CACHE)
```

### Setup: restore the cached build from Google Drive (first run clones and builds, then saves it)

```python
import os, sys, glob, subprocess

REBUILD = False   # True = ignore the cached build, pull the latest Unsloth, and rebuild (then re-save it)

BUILD_DIR = "/content/drive/MyDrive/llm_models/builds"
os.makedirs(BUILD_DIR, exist_ok=True)
HOME = os.path.expanduser("~")
TARBALL = f"{BUILD_DIR}/unsloth_studio_py{sys.version_info.major}{sys.version_info.minor}.tar"
# Everything setup.sh produces: the cloned repo (with the built frontend), plus venvs, node and llama.cpp under ~/.unsloth
PATHS = ["/content/unsloth", f"{HOME}/.unsloth"] + [p for p in (f"{HOME}/.local/share/uv",) if os.path.exists(p)]

def ready():
    pys = glob.glob(f"{HOME}/.unsloth/studio/*/bin/python")
    return (os.path.exists("/content/unsloth/studio/frontend/dist/index.html") and bool(pys)
            and subprocess.run([pys[0], "-c", "import sys"], capture_output=True).returncode == 0)

if not REBUILD and os.path.exists(TARBALL):
    print("Restoring cached build from Drive...")
    subprocess.run(["tar", "-xf", TARBALL, "-C", "/"], check=True)

if ready():
    print("Cached build is ready.")
else:
    print("No usable cached build; cloning and building (this is the slow part)...")
    subprocess.run("rm -rf /content/unsloth", shell=True)
    !git clone --depth 1 --branch main https://github.com/unslothai/unsloth.git /content/unsloth
    !cd /content/unsloth && chmod +x studio/setup.sh && ./studio/setup.sh --local
    print("Saving the build to Drive for next time...")
    rel = [p.lstrip("/") for p in PATHS]
    subprocess.run(["tar", "-cf", TARBALL + ".part", "-C", "/"] + rel, check=True)
    os.rename(TARBALL + ".part", TARBALL)
    print("Saved:", TARBALL, f"({os.path.getsize(TARBALL)/1e9:.1f} GB)")
os.chdir("/content/unsloth")
```
